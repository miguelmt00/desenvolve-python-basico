# Resolução dos Exercícios em Python

Este notebook contém a resolução dos exercícios de fluxograma e repetição em Python.

## 1. Fluxograma — Verificar se x é maior que 5

````python
x = int(input("Digite o valor de x: "))

if x > 5:
    print("Maior que 5")

print("Fim")
````

## 2. Fluxograma — Contador

````python
n = int(input("Digite o valor de n: "))

cont = 0

while n < cont:
    cont = cont + 1
    print(cont)

print("Fim")
````

> Observação: seguindo exatamente o fluxograma, a condição é `n < cont`. Como `cont` começa em 0, esse laço só executa se `n` for menor que 0.

## 3. Fluxograma — Média e situação do aluno

````python
n1 = float(input("Digite a primeira nota: "))
n2 = float(input("Digite a segunda nota: "))
n3 = float(input("Digite a terceira nota: "))

m = (n1 + n2 + n3) / 3

if m >= 60:
    print("Aprovado")
elif m >= 40:
    print("Recuperação")
else:
    print("Reprovado")

print("Fim")
````

## 4. Fluxograma — Encontrar o maior valor

````python
n = int(input("Digite a quantidade de números: "))

maior = 0

while n > 0:
    x = int(input("Digite um número: "))

    if x > maior:
        maior = x

    n = n - 1

print(maior)
````

## 5. Média de idade dos respondentes

````python
n = int(input("Digite a quantidade de respondentes: "))

soma = 0

for i in range(n):
    idade = int(input("Digite a idade: "))
    soma = soma + idade

media = soma / n

print(f"Média das idades: {media:.2f}")
````

## 6. Beecrowd 1094 — Experiências

````python
n = int(input())

total = 0
total_coelhos = 0
total_ratos = 0
total_sapos = 0

for i in range(n):
    quantidade, tipo = input().split()
    quantidade = int(quantidade)

    total = total + quantidade

    if tipo == "C":
        total_coelhos = total_coelhos + quantidade
    elif tipo == "R":
        total_ratos = total_ratos + quantidade
    elif tipo == "S":
        total_sapos = total_sapos + quantidade

percentual_coelhos = (total_coelhos / total) * 100
percentual_ratos = (total_ratos / total) * 100
percentual_sapos = (total_sapos / total) * 100

print(f"Total: {total} cobaias")
print(f"Total de coelhos: {total_coelhos}")
print(f"Total de ratos: {total_ratos}")
print(f"Total de sapos: {total_sapos}")
print(f"Percentual de coelhos: {percentual_coelhos:.2f} %")
print(f"Percentual de ratos: {percentual_ratos:.2f} %")
print(f"Percentual de sapos: {percentual_sapos:.2f} %")
````
