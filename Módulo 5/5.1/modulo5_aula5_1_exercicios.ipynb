```python
num1=float(input("Digite o primeiro número: "))
num2=float(input("Digite o segundo número: "))
diferenca=abs(num1-num2)
print(f"A diferença absoluta entre os números é: {round(diferenca,2)}")
```

```python
import random, math
n=int(input("Digite a quantidade de números aleatórios: "))
soma=0
for i in range(n):
    soma+=random.randint(0,100)
print(f"Soma dos valores: {soma}")
print(f"Raiz quadrada da soma: {math.sqrt(soma):.2f}")
```

```python
import random
numero_secreto=random.randint(1,10)
while True:
    palpite=int(input("Adivinhe o número entre 1 e 10: "))
    if palpite>numero_secreto:
        print("Muito alto, tente novamente!")
    elif palpite<numero_secreto:
        print("Muito baixo, tente novamente!")
    else:
        print(f"Correto! O número é {numero_secreto}.")
        break
```
