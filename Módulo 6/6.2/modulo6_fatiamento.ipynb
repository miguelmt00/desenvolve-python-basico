## Questão 1

Escreva um script em Python que solicita do usuário uma quantidade indefinida de números inteiros (com pelo menos 4 valores), os armazena em uma lista e, usando fatiamento de listas, imprima:

- A lista original
- Os 3 primeiros elementos
- Os 2 últimos elementos
- A lista invertida
- Os elementos de índice par
- Os elementos de índice ímpar

```python
numeros = []

print('Digite pelo menos 4 números inteiros. Digite 0 para finalizar.')

while True:
    n = int(input())
    if n == 0 and len(numeros) >= 4:
        break
    numeros.append(n)

print('\nLista original:', numeros)
print('3 primeiros elementos:', numeros[:3])
print('2 últimos elementos:', numeros[-2:])
print('Lista invertida:', numeros[::-1])
print('Índices pares:', numeros[::2])
print('Índices ímpares:', numeros[1::2])
```

## Questão 2

Dada uma lista de URLs que começam com 'www.' e terminam com '.com', utilize fatiamento para criar uma lista contendo apenas os nomes principais dos domínios.

```python
urls = ['www.google.com', 'www.gmail.com', 'www.github.com', 'www.reddit.com', 'www.yahoo.com']

dominios = [url[4:-4] for url in urls]

print('URLs:', urls)
print('Domínios:', dominios)
```

## Questão 3

Crie uma lista com 20 elementos entre -10 e 10 gerados aleatoriamente. Encontre o maior intervalo contínuo de números negativos e remova-o utilizando o operador del. Imprima a lista antes e depois da remoção.

```python
import random

lista = [random.randint(-10, 10) for _ in range(20)]

print('Original:', lista)

inicio = fim = -1
maior_inicio = maior_fim = -1
maior_tamanho = 0

for i in range(len(lista)):
    if lista[i] < 0:
        if inicio == -1:
            inicio = i
        fim = i
    else:
        if inicio != -1:
            tamanho = fim - inicio + 1
            if tamanho > maior_tamanho:
                maior_tamanho = tamanho
                maior_inicio = inicio
                maior_fim = fim
            inicio = fim = -1

if inicio != -1:
    tamanho = fim - inicio + 1
    if tamanho > maior_tamanho:
        maior_inicio = inicio
        maior_fim = fim

if maior_inicio != -1:
    del lista[maior_inicio:maior_fim+1]

print('Editada:', lista)
```
