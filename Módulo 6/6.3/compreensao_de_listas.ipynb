## Questão 1

Escreva scripts usando compreensão de listas para:

1. Todos os números pares entre 20 e 50.
2. Os quadrados dos valores da lista [1,2,3,4,5,6,7,8,9].
3. Todos os números entre 1 e 100 divisíveis por 7.
4. Para os valores em range(0,30,3), escrever 'par' ou 'ímpar'.

```python
pares = [x for x in range(20, 51) if x % 2 == 0]
quadrados = [x**2 for x in [1,2,3,4,5,6,7,8,9]]
divisiveis_por_7 = [x for x in range(1,101) if x % 7 == 0]
paridade = ['par' if x % 2 == 0 else 'ímpar' for x in range(0,30,3)]

print('Pares:', pares)
print('Quadrados:', quadrados)
print('Divisíveis por 7:', divisiveis_por_7)
print('Paridade:', paridade)
```

## Questão 2

Solicite uma frase do usuário e usando compreensão de listas imprima:

- A lista de vogais ordenada alfabeticamente.
- A lista de consoantes sem espaços em branco.

```python
frase = input('Digite uma frase: ')

vogais = sorted([c.lower() for c in frase if c.lower() in 'aeiou'])
consoantes = [c for c in frase if c.isalpha() and c.lower() not in 'aeiou']

print('Vogais:', vogais)
print('Consoantes:', consoantes)
```

## Questão 3

Reescreva o código para construir a lista pagamentos utilizando compreensão de listas.

```python
horas_trabalhadas = [40, 37, 45, 40, 40, 48]
ganho_por_hora = 20
hora_extra = 25

pagamentos = [
    ganho_por_hora * min(hora, 40) + hora_extra * max(0, hora - 40)
    for hora in horas_trabalhadas
]

print(pagamentos)
```

## Questão 4

Reescreva o código para construir a lista aprovados utilizando compreensão de listas.

```python
alunos = ['Maria', 'Jose', 'Carla', 'Sol']
notas = [35, 50, 20, 80]

aprovados = [alunos[i] for i in range(len(notas)) if notas[i] >= 60]

print('Aprovados:', aprovados)
```
